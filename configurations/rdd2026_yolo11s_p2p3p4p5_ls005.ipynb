```python
from ultralytics import YOLO
model_name = "yolo11s_p2p3p4p5"
ls = 0.05
# Build a YOLOv11l model from pretrained weight
model = YOLO(f"{model_name}.yaml")
model.info()
```

```python
# Load pretrained weights
model = model.load("yolo11s.pt")
```

```python
def format_float(value):
    # Multiply by 100 to shift the decimal and convert to int
    scaled_value = int(value * 100)
    # Format as a 3-digit zero-padded string
    return f"{scaled_value:03d}"

```

```python
# Training.
results = model.train(
   data='/home/pham/Documents/SHSU/Researching/roaddamagedetector/rddallcountriespotholesfortrainyolov11.yaml',
   imgsz=640,
   epochs=150,
   batch=32,
   cls_label_smoothing=ls,
   name=f"{model_name}_ls{format_float(ls)}")
```
